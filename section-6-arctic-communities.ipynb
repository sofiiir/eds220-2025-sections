```python
# Make an if/else statemnt to check geometries
if len(df.geom_type.unique()) != 1:
    print(f"Multiple feature types: {df.geom_type.unique()}")
else:
    print(f"All feature are:{df.geom_type.unique()}")
```

```python
# Make an if else statement function 
def check_polygons(df1):
    if len(df1.geom_type.unique()) != 1:
        print(f"Multiple feature types: {df.geom_type.unique()}")
    else:
        print(f"All feature are:{df.geom_type.unique()}")
```

```python
# Run the function on the dataframe
check_polygons(df)
```

## Explode polygons

```python
# Explode multi-polygons and reset index
df = df.explode(column = 'geometry', index_parts = False).reset_index()
```

```python
df
```

```python
# Run the check-polygons function on our exploded df
check_polygons(df)
```

## Compute minimum y-coordinate for polygons

```python
df.iloc[0,]
```

```python
type(df.iloc[0,])
```

```python
geometry_1 = df.iloc[0,].geometry
```

```python
geometry_1.bounds[1]
```

```python
def miny(row):
    return row.geometry.bounds[1]
```

```python
miny(df.iloc[0,])
```

```python
df['miny']= df.apply(miny, axis = 1)
```

## Filter, update CRS, and reproduce map

```python
arctic = df[df['miny'] >40]
```

```python
arctic= arctic.to_crs('epsg:3413')
```

```python
fig, ax = plt.subplots(figsize=(8, 6))

# Remove the axis for a cleaner map
ax.axis('off')

# Create choropleth map of communities
# Plot with refined color and edge style
arctic.plot(
    ax=ax
    
)

# Add title and subtitle for better context
ax.set_title('Distribution of Arctic Communities', fontsize=18, weight='bold', pad=15)

plt.show()
```
