```python
# Make an if/else statemnt to check geometries
if len(df.geom_type.unique()) != 1:
    print(f"Multiple feature types: {df.geom_type.unique()}")
else:
    print(f"All feature are:{df.geom_type.unique()}")
```

```python
# Make an if else statement function 
def check_polygons(df1):
    if len(df1.geom_type.unique()) != 1:
        print(f"Multiple feature types: {df.geom_type.unique()}")
    else:
        print(f"All feature are:{df.geom_type.unique()}")
```

```python
check_polygons(df)
```
