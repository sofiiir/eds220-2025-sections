# Water conflicts in the Colorado River Basin 

### Archive exploration 
1. Where was the data collected from??
 The data was collected via two methods. The first method was researchers observing meetings at water control boards and organizations in the Colorado River Basin. This data was analyzed for the ,major themes discussed at the meetings. The second method of data collection was looking at the archives of water-related events in the basin from newspapers.
2. During what time frame were the observations in the dataset collected?
The data is from 2005-2021.
3. What was the author’s perceived value of this dataset?
 The author believes this dataset is useful background information for the USGS which producing new water availability assessment reports.
